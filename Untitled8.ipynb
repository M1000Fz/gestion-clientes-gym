```python
import psycopg2
import ipywidgets as widgets
from IPython.display import display, clear_output


DATABASE_URL = "postgresql://neondb_owner:npg_cpbBGC6hKV8x@ep-twilight-unit-aejcmcmn-pooler.c-2.us-east-2.aws.neon.tech/neondb?sslmode=require&channel_binding=require"

# Función auxiliar para ejecutar consultas SQL y evitar repetir código
def ejecutar_bd(consulta, datos=None, retornar_filas=False):
    conexion = None
    try:
        conexion = psycopg2.connect(DATABASE_URL)
        cursor = conexion.cursor()
        cursor.execute(consulta, datos)

        if retornar_filas:
            resultado = cursor.fetchall()
            return resultado

        conexion.commit()
        return True
    except Exception as e:
        return f"Error: {e}"
    finally:
        if conexion:
            cursor.close()
            conexion.close()

# ==========================================
# PESTAÑA 1: REGISTRAR CLIENTE
# ==========================================
input_nombre = widgets.Text(description="Nombre:")
input_apellido = widgets.Text(description="Apellido:")
input_email = widgets.Text(description="Email:")
input_telefono = widgets.Text(description="Teléfono:")
input_membresia = widgets.Dropdown(options=['Mensual', 'Trimestral', 'Anual', 'VIP'], value='Mensual', description='Membresía:')
boton_guardar = widgets.Button(description='Guardar Cliente', button_style='success', icon='check')
salida_registro = widgets.Output()

def guardar_datos(b):
    with salida_registro:
        clear_output()
        if not input_nombre.value or not input_apellido.value:
            print("⚠️ El nombre y apellido son obligatorios.")
            return

        consulta = """
            INSERT INTO clientes (nombre, apellido, email, telefono, tipo_membresia)
            VALUES (%s, %s, %s, %s, %s);
        """
        datos = (input_nombre.value, input_apellido.value, input_email.value, input_telefono.value, input_membresia.value)

        resultado = ejecutar_bd(consulta, datos)
        if resultado is True:
            print(f"✅ Cliente {input_nombre.value} registrado exitosamente.")
            input_nombre.value = input_apellido.value = input_email.value = input_telefono.value = ""
            input_membresia.value = "Mensual"
        else:
            print("❌ Ocurrió un error:", resultado)

boton_guardar.on_click(guardar_datos)
pestaña_registro = widgets.VBox([input_nombre, input_apellido, input_email, input_telefono, input_membresia, boton_guardar, salida_registro])


# ==========================================
# PESTAÑA 2: VISUALIZAR CLIENTES
# ==========================================
boton_actualizar_lista = widgets.Button(description='Refrescar Lista', button_style='info', icon='refresh')
salida_lista = widgets.Output()

def mostrar_lista(b):
    with salida_lista:
        clear_output()
        consulta = "SELECT id, nombre, apellido, email, telefono, tipo_membresia, estado FROM clientes ORDER BY id ASC;"
        filas = ejecutar_bd(consulta, retornar_filas=True)

        if isinstance(filas, str) and filas.startswith("Error"):
            print(filas)
        elif not filas:
            print("No hay clientes registrados en la base de datos.")
        else:
            print("LISTA DE CLIENTES:")
            print("=" * 80)
            for fila in filas:
                texto_cliente = f"ID: {fila[0]} | Nombre: {fila[1]} {fila[2]} | Correo: {fila[3]} | Tel: {fila[4]} | Plan: {fila[5]} | Estado: {fila[6]}"
                print(texto_cliente)
                print("-" * 80)

boton_actualizar_lista.on_click(mostrar_lista)
pestaña_visualizar = widgets.VBox([boton_actualizar_lista, salida_lista])


# ==========================================
# PESTAÑA 3: MODIFICAR / ELIMINAR
# ==========================================
edit_id = widgets.IntText(description="ID Cliente:")
boton_buscar = widgets.Button(description="Buscar Datos", button_style='info', icon='search')

edit_nombre = widgets.Text(description="Nombre:")
edit_apellido = widgets.Text(description="Apellido:")
edit_email = widgets.Text(description="Email:")
edit_telefono = widgets.Text(description="Teléfono:")
edit_membresia = widgets.Dropdown(options=['Mensual', 'Trimestral', 'Anual', 'VIP'], description='Membresía:')

boton_actualizar = widgets.Button(description="Actualizar", button_style='warning', icon='edit')
boton_eliminar = widgets.Button(description="Eliminar", button_style='danger', icon='trash')
salida_edicion = widgets.Output()

def buscar_cliente(b):
    with salida_edicion:
        clear_output()
        consulta = "SELECT nombre, apellido, email, telefono, tipo_membresia FROM clientes WHERE id = %s;"
        filas = ejecutar_bd(consulta, (edit_id.value,), retornar_filas=True)

        if isinstance(filas, str) and filas.startswith("Error"):
            print(filas)
        elif not filas:
            print(f"⚠️ No se encontró ningún cliente con el ID {edit_id.value}.")
        else:
            datos = filas[0]
            edit_nombre.value, edit_apellido.value = datos[0], datos[1]
            # Manejar posibles valores nulos en la BD
            edit_email.value = datos[2] if datos[2] else ""
            edit_telefono.value = datos[3] if datos[3] else ""
            edit_membresia.value = datos[4] if datos[4] in ['Mensual', 'Trimestral', 'Anual', 'VIP'] else 'Mensual'
            print("✅ Datos cargados. Ahora puedes modificarlos o eliminar el registro.")

def actualizar_cliente(b):
    with salida_edicion:
        clear_output()
        consulta = """
            UPDATE clientes
            SET nombre = %s, apellido = %s, email = %s, telefono = %s, tipo_membresia = %s
            WHERE id = %s;
        """
        datos = (edit_nombre.value, edit_apellido.value, edit_email.value, edit_telefono.value, edit_membresia.value, edit_id.value)
        resultado = ejecutar_bd(consulta, datos)
        if resultado is True:
            print(f"✅ Cliente ID {edit_id.value} actualizado correctamente.")
        else:
            print("❌ Error al actualizar:", resultado)

def eliminar_cliente(b):
    with salida_edicion:
        clear_output()
        consulta = "DELETE FROM clientes WHERE id = %s;"
        resultado = ejecutar_bd(consulta, (edit_id.value,))
        if resultado is True:
            print(f"🗑️ Cliente ID {edit_id.value} eliminado de la base de datos.")
            edit_nombre.value = edit_apellido.value = edit_email.value = edit_telefono.value = ""
        else:
            print("❌ Error al eliminar:", resultado)

boton_buscar.on_click(buscar_cliente)
boton_actualizar.on_click(actualizar_cliente)
boton_eliminar.on_click(eliminar_cliente)

caja_botones_edicion = widgets.HBox([boton_actualizar, boton_eliminar])
pestaña_editar = widgets.VBox([
    widgets.HTML("<b>Ingresa el ID del cliente que deseas modificar o borrar:</b>"),
    widgets.HBox([edit_id, boton_buscar]),
    widgets.HTML("<hr>"),
    edit_nombre, edit_apellido, edit_email, edit_telefono, edit_membresia,
    caja_botones_edicion,
    salida_edicion
])


# ==========================================
# RENDERIZAR INTERFAZ CON PESTAÑAS
# ==========================================
panel_principal = widgets.Tab(children=[pestaña_registro, pestaña_visualizar, pestaña_editar])
panel_principal.set_title(0, '1. Registrar')
panel_principal.set_title(1, '2. Ver Clientes')
panel_principal.set_title(2, '3. Editar / Eliminar')

display(widgets.HTML("<h3>⚙️ Sistema de Gestión - Gimnasio</h3>"))
display(panel_principal)
```
